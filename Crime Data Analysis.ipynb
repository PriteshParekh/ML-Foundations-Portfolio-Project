```python
# Each imported dataset have identical structure as seen above

# Next step: to create a master dataframe to store the years crime data
```

```python
jan_2025["month"] = "2025-01"
feb_2025["month"] = "2025-02"
mar_2025["month"] = "2025-03"
apr_2025["month"] = "2025-04"
may_2025["month"] = "2025-05"
jun_2025["month"] = "2025-06"
jul_2025["month"] = "2025-07"
aug_2025["month"] = "2025-08"
sep_2025["month"] = "2025-09"
oct_2025["month"] = "2025-10"
nov_2025["month"] = "2025-11"
dec_2025["month"] = "2025-12"

master_2025 = pd.concat([jan_2025, feb_2025, mar_2025, apr_2025, may_2025, jun_2025, jul_2025, aug_2025, sep_2025, oct_2025, nov_2025, dec_2025], ignore_index = True)

print(master_2025.shape)
print(master_2025.head())
```

```python
# Successfully concatenated each data sheet into a master sheet representing the year.

# Next Step: check 3 random months to see if the length matchs the original month_df length. If so, its safe to confirm the concatenation
```

```python
def check_df_length(original_datasheet, month):
    counter = 0
    for x in master_2025['month']:
        if x == month:
            counter += 1
    if counter == len(original_datasheet['month']):
        print(True)
    else:
        print(False)
```

```python
check_df_length(mar_2025, '2025-03')
check_df_length(sep_2025, '2025-09')
check_df_length(dec_2025, '2025-12')
```

```python
# As the returned value is True for all 3 cases, this means that the length of the (total) months found in the master is the same as the total length in the original. Therfore as the implementation was the same across all months, we can confindantley assume the concatination was successful.

# Next Step: Clean the dataset
```

```python
print(master_2025.head())
print(master_2025.columns)
```

```python
# Upon initial review, some columns do not seem to offer any insights, such as; 
# Crime ID: not useful in this case,
# Longitude, Latitude: Unable to convert data into location, additionally location is provided
# context: N/a
# LSOA code, Reported by: not useful in this case
# month is a duplicate of Month
```

```python
list_to_drop = ["Crime ID", "Longitude", "Latitude", "LSOA code", "Context", "month"]
dropped_master_2025 = master_2025.drop(list_to_drop, axis=1)

print(dropped_master_2025.columns)
print(dropped_master_2025.sample(12))
```

```python
print(dropped_master_2025.isna().sum())
```

```python
dropped_master_2025 = dropped_master_2025.dropna(subset=["LSOA name", "Last outcome category"])
print(dropped_master_2025.isna().sum())
```
