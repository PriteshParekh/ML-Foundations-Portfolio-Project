```python
# Each imported dataset have identical structure as seen above

# Next step: to create a master dataframe to store the years crime data
```

```python
jan_2025["month"] = "2025-01"
feb_2025["month"] = "2025-02"
mar_2025["month"] = "2025-03"
apr_2025["month"] = "2025-04"
may_2025["month"] = "2025-05"
jun_2025["month"] = "2025-06"
jul_2025["month"] = "2025-07"
aug_2025["month"] = "2025-08"
sep_2025["month"] = "2025-09"
oct_2025["month"] = "2025-10"
nov_2025["month"] = "2025-11"
dec_2025["month"] = "2025-12"

master_2025 = pd.concat([jan_2025, feb_2025, mar_2025, apr_2025, may_2025, jun_2025, jul_2025, aug_2025, sep_2025, oct_2025, nov_2025, dec_2025], ignore_index = True)

print(master_2025.shape)
print(master_2025.head())
```

```python
# Successfully concatenated each data sheet into a master sheet representing the year.

# Next Step: check 3 random months to see if the length matchs the original month_df length. If so, its safe to confirm the concatenation
```

```python
def check_df_length(original_datasheet, month):
    counter = 0
    for x in master_2025['month']:
        if x == month:
            counter += 1
    if counter == len(original_datasheet['month']):
        print(True)
    else:
        print(False)
```

```python
check_df_length(mar_2025, '2025-03')
check_df_length(sep_2025, '2025-09')
check_df_length(dec_2025, '2025-12')
```

```python
# As the returned value is True for all 3 cases, this means that the length of the (total) months found in the master is the same as the total length in the original. Therfore as the implementation was the same across all months, we can confindantley assume the concatination was successful.

# Next Step: Clean the dataset
```

```python
print(master_2025.head())
print(master_2025.columns)
```

```python
# Upon initial review, some columns do not seem to offer any insights, such as; 
# Crime ID: not useful in this case,
# Longitude, Latitude: Unable to convert data into location, additionally location is provided
# context: N/a
# LSOA code, Reported by: not useful in this case
# month is a duplicate of Month
```

```python
list_to_drop = ["Crime ID", "Longitude", "Latitude", "LSOA code", "Context", "month"]
dropped_master_2025 = master_2025.drop(list_to_drop, axis=1)

print(dropped_master_2025.columns)
print(dropped_master_2025.sample(12))
```

```python
print(dropped_master_2025.isna().sum())
```

```python
dropped_master_2025 = dropped_master_2025.dropna(subset=["LSOA name", "Last outcome category"])
print(dropped_master_2025.isna().sum())
```

```python
print(dropped_master_2025.duplicated().sum())
print(dropped_master_2025.dtypes)
print(dropped_master_2025.nunique())
```

```python
# This result indicates there are multiple duplicated records, however this can be interpreted as "same crime in the same location".
# As many people can commit many crimes, it is reasonable to assume similar crimes can be committed in the same area. Duplicates are subject for removal.
```

```python
cleaned_master_2025 = dropped_master_2025[dropped_master_2025.duplicated(keep=False)]
cleaned_master_2025.head()
```

```python
# 1. b. Are there duplicate records in the dataset?
```

```python
print(
    "There are {} duplicated records found in the dataset. "
    "Due to the nature of the dataset, the duplicated records have been kept under the assumption that the same crime can be committed in the same area/location by the same or another offender\nDuplicates are subject for removal after review."
    .format(dropped_master_2025.duplicated().sum())
)
```

```python
duplicate_rows = dropped_master_2025[dropped_master_2025.duplicated(keep=False)]
duplicate_counts = duplicate_rows.value_counts().reset_index(name="count")

duplicate_counts["record"] = duplicate_counts.astype(str).agg(" | ".join, axis=1)

plt.figure(figsize=(14, 8))
plt.barh(duplicate_counts["record"].head(10), duplicate_counts["count"].head(10))
plt.title("Top Duplicated Crime Records")
plt.xlabel("Number of Times Duplicated")
plt.ylabel("Duplicated Record")
plt.tight_layout()
plt.show()
```

```python
# Upon review of the chart, it is highly unlikely for crimes such as "Theft", "Violence and Sexual Offences" and "Shoplifting" to be committed at the same location, within the same 100> times. Therefore after review, duplicates should be removed and only one instance should be retained.
```

```python
# 3) Monthly trends
```

```python
# 3. a. How many crimes were recorded each month?

# 3. b. Which month had the highest crime count?

# 3. c. Which month had the lowest crime count?
```

```python
cleaned_master_2025["Month"] = pd.to_datetime(cleaned_master_2025["Month"])
cleaned_master_2025["Month_Name"] = cleaned_master_2025["Month"].dt.strftime("%b")

month_order = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
               "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

monthly_crimes_named = cleaned_master_2025.groupby("Month_Name").size().reindex(month_order).astype(int)

print("Crimes recorded each month:")
for month, total in monthly_crimes_named.items():
    print("{}: {}".format(month, total))

highest_m = monthly_crimes_named.idxmax()
highest_v = monthly_crimes_named.max()

lowest_m = monthly_crimes_named.idxmin()
lowest_v = monthly_crimes_named.min()

print("\n")
print("{} has the most recorded crime at {}".format(highest_m, highest_v))
print("{} has the least recorded crime at {}".format(lowest_m, lowest_v))
```

```python
cleaned_master_2025["Month"] = pd.to_datetime(cleaned_master_2025["Month"])
cleaned_master_2025["Month_Name"] = cleaned_master_2025["Month"].dt.strftime("%b")

month_order = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
               "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

monthly_crimes = cleaned_master_2025.groupby("Month_Name").size().reindex(month_order)

plt.figure(figsize=(10, 6))
monthly_crimes.plot(kind="bar", color="steelblue")
plt.title("Total Number of Crimes Committed Each Month")
plt.xlabel("Month")
plt.ylabel("Number of Crimes")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# 3. d. How many unique crime types appear in each month?
```

```python
summary = (
    cleaned_master_2025.groupby("Month_Name")["Crime type"]
    .agg(
        Unique_Crime_Types=lambda x: sorted(x.unique().tolist()),
        Count="nunique"
    )
    .reset_index()
)

month_order = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
               "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

summary = summary.set_index("Month_Name").reindex(month_order).reset_index()
print(summary)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.bar(summary["Month_Name"], summary["Count"], color="steelblue")
plt.title("Unique Crime Types in Each Month")
plt.xlabel("Month")
plt.ylabel("Number of Unique Crime Types")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Each month saw 13 different crimes (all) committed.
```

```python
monthly_crime_counts = (
    cleaned_master_2025.groupby(["Month_Name", "Crime type"])
    .size()
    .reset_index(name="Count")
)

month_order = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
               "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

monthly_crime_counts["Month_Name"] = pd.Categorical(
    monthly_crime_counts["Month_Name"],
    categories=month_order,
    ordered=True
)

monthly_crime_counts = monthly_crime_counts.sort_values(["Month_Name", "Count"], ascending=[True, False])
print(monthly_crime_counts)
```

```python
# 3. e. Which month has the widest variety of crime types?
```

```python
pivot = monthly_crime_counts.pivot(index="Month_Name", columns="Crime type", values="Count").fillna(0)
pivot = pivot.reindex(month_order)

pivot.plot(kind="bar", figsize=(14, 7))
plt.title("Crime Types Committed Each Month")
plt.xlabel("Month")
plt.ylabel("Number of Crimes")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Violence and Sexual Offences was committed the most for every month of the year, followed by shoplifting. Other theft and Public order was reported similarly across all the months, however Public order rose from March to October, then reducing slightly in November and December. Criminal damage was reported almost consistently across all the months.
```

```python
# 5) Geographic analysis
```

```python
# 5. a. Which location has the highest number of recorded crimes?
```

```python
reporter_counts = cleaned_master_2025["Reported by"].value_counts()

top_reporter = reporter_counts.idxmax()
top_reporter_count = reporter_counts.max()

top_row = cleaned_master_2025[cleaned_master_2025["Reported by"] == top_reporter].iloc[0]
top_location = top_row["Location"]
top_falls_within = top_row["Falls within"]

print(
    "The reporter '{}' recorded the highest number of crimes at {}. "
    "The top matching location was '{}' in {}, with {} recorded crimes."
    .format(top_reporter, top_reporter_count, top_location, top_falls_within, top_reporter_count)
)
```

```python
falls_within_counts = cleaned_master_2025["Falls within"].value_counts()

plt.figure(figsize=(10, 6))
falls_within_counts.head(10).sort_values().plot(kind="barh")
plt.title("Top 10 Falls Within Areas by Recorded Crimes")
plt.xlabel("Number of Crimes")
plt.ylabel("Falls within")
plt.tight_layout()
plt.show()
```
